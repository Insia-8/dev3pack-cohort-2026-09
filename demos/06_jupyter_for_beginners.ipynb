# Demo 6 — Jupyter for beginners

**Never used a notebook before? Start here.** Fifteen minutes, and the rest of the
course stops being mysterious.

A notebook is a list of **cells**. A cell holds either text (like this one) or
code. You run a code cell and its output appears underneath it, inside the same
document.

**The two things to know before anything else:**

| | |
|---|---|
| **Run a cell** | Click it, then press **Shift + Enter** |
| **Something is stuck** | Menu: **Kernel → Restart Kernel and Run All Cells** |

Nothing in this notebook can break anything. Change a line, run it again, see what
happens. That is the whole point of a notebook.

## 1. Your first cell

Click the cell below once, then press **Shift + Enter**.

You will see `[ ]` on its left turn into `[*]` while it runs, then into a number.
**That number is the order cells ran in**, and it matters more than it looks.

````python
print("Hello. You just ran code.")
2 + 2
````

Two things happened:

- `print(...)` wrote a line.
- `2 + 2` was the **last line**, so the notebook showed its value: `4`.

A cell shows the value of its last line automatically. That is why `4` appears
without a `print`.

## 2. Cells share memory, and they remember

Anything you name in one cell exists in the next one. Run these two in order.

````python
name = "Dev3Pack"
session = 4
````

````python
print(f"{name}, session {session}")
````

**This is where the first real confusion comes from.** Memory belongs to the
*kernel*, the Python process behind the notebook, not to the file. So:

- If you **change a cell and do not re-run it**, the old value is still in memory.
- If you **run cells out of order**, the result depends on the order you ran
  them, not on the order you see.

The numbers in `[ ]` show what really happened. Read them when something makes
no sense.

## 3. See it go wrong on purpose

The cell below uses a name that does not exist. Run it: the notebook shows an
error and **keeps going** — nothing is lost, nothing is broken.

(We catch the error and print it, so this demo still runs top to bottom for
everyone. In your own notebooks the red block appears by itself.)

````python
import traceback

try:
    print(total_price)                    # never defined anywhere
except NameError:
    print(traceback.format_exc().strip())
````

**How to read any Python error, in three steps:**

1. **Read the last line first.** It names what went wrong:
   `NameError: name 'total_price' is not defined`.
2. **Then look just above it** for the line of *your* code that caused it.
3. **Fix that one thing** and run the cell again.

The four errors you will meet this month:

| Error | Usually means |
|---|---|
| `NameError` | A typo, or you did not run the cell that creates it |
| `ModuleNotFoundError` | Something is not installed: run `uv sync` in the terminal |
| `FileNotFoundError` | The path is wrong, or you are in a different folder |
| `KeyError` | You asked a dictionary for a key it does not have |

## 4. Where am I, and what can I see?

A notebook runs in a folder. Every relative path starts from there.

````python
from pathlib import Path

here = Path.cwd()
print("this notebook runs in:", here)
print()
print("the first few things next to it:")
for item in sorted(here.iterdir())[:6]:
    print("  ", item.name)
````

## 5. The buttons that matter

Everything else can wait.

| What you want | How |
|---|---|
| Run this cell, move to the next | **Shift + Enter** |
| Run this cell, stay here | **Ctrl + Enter** |
| Add a cell below | **Esc**, then **B** |
| Delete a cell | **Esc**, then **D D** (press D twice) |
| Undo | **Ctrl + Z** inside a cell; **Esc** then **Z** for a deleted cell |
| Stop a cell that runs forever | The **■** stop button, or **Kernel → Interrupt** |
| Start completely fresh | **Kernel → Restart Kernel and Run All Cells** |
| Save | **Ctrl + S**. Do this before you hand work in |

**Esc** leaves the cell so the keys act on the notebook. **Enter** goes back into
the cell. That single idea explains most keyboard confusion.

## 6. The one habit that saves the most time

Before you ask anybody for help, and before you hand work in:

> **Kernel → Restart Kernel and Run All Cells**

It throws away the memory and runs everything from the top. If it passes, your
notebook really works. If it fails, you have just found a cell that only worked
because of something you ran earlier and then changed.

**On this course it matters twice over:** `bootcamp check` runs your notebook in a
fresh process, and `bootcamp submit` reads the file on disk. A notebook that only
works in your current session is a notebook that fails both.
